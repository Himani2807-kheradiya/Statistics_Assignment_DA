```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

Task 1: Create a CSV file named purchase_data.csv

```python
data = {
    "customer_id": range(1, 16),
    "age_group": [
        "18-25", "18-25", "18-25", "18-25", "18-25",
        "26-35", "26-35", "26-35", "26-35", "26-35",
        "36-50", "36-50", "36-50", "36-50", "36-50"
    ],
    "purchase_value": [
        250, 300, 280, 320, 270,
        400, 420, 380, 450, 410,
        500, 520, 480, 550, 510
    ]
}

df = pd.DataFrame(data)

# Save the data into a CSV file
df.to_csv("purchase_data.csv", index=False)

# Display the dataset
print(df)
```

Task 2: Perform One-Way ANOVA using pandas and scipy.stats

```python
df = pd.read_csv("purchase_data.csv")

group_18_25 = df[df["age_group"] == "18-25"]["purchase_value"]
group_26_35 = df[df["age_group"] == "26-35"]["purchase_value"]
group_36_50 = df[df["age_group"] == "36-50"]["purchase_value"]

f_statistic, p_value = stats.f_oneway(
    group_18_25,
    group_26_35,
    group_36_50
)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
```

Explanation (Markdown cell):

One-Way ANOVA is used to compare the average purchase values of three independent age groups.

First, we read the CSV file using Pandas. Then, we separated the purchase values into three groups: 18-25, 26-35, and 36-50.

We used stats.f_oneway() from SciPy to perform the ANOVA test.

The function returns two important results:
- F-statistic: Measures the variation between group means compared with the variation within the groups.
- P-value: Helps determine whether the differences between group means are statistically significant.
The calculated F-statistic is approximately 94.6473, and the p-value is approximately 0.0000000449.

Task 3: Interpret the ANOVA results

The One-Way ANOVA test checks whether the average purchase values are significantly different across the three age groups.

The p-value is approximately 0.0000000449, which is less than the significance level of 0.05. Therefore, we reject the null hypothesis.

Conclusion: There is a statistically significant difference in average purchase values between at least two age groups. This suggests that customer age groups may have different purchasing patterns.

Task 4: Calculate between-group variance and within-group variance

```python
grand_mean = df["purchase_value"].mean()

number_of_groups = df["age_group"].nunique()
total_observations = len(df)

ss_between = 0

for group in df["age_group"].unique():
    group_data = df[df["age_group"] == group]["purchase_value"]
    group_mean = group_data.mean()
    group_size = len(group_data)

    ss_between += group_size * (group_mean - grand_mean) ** 2

ss_within = 0

for group in df["age_group"].unique():
    group_data = df[df["age_group"] == group]["purchase_value"]
    group_mean = group_data.mean()

    ss_within += ((group_data - group_mean) ** 2).sum()

df_between = number_of_groups - 1
df_within = total_observations - number_of_groups

between_group_variance = ss_between / df_between
within_group_variance = ss_within / df_within

print("Grand Mean:", grand_mean)
print("Between-Group Variance (MSB):", between_group_variance)
print("Within-Group Variance (MSW):", within_group_variance)
```

Explanation (Markdown cell):

In ANOVA, variance is used to understand the differences between customer groups and the variation within each group.

1. Between-group variance (MSB):

It measures how much the average purchase values differ between the three age groups.

A higher value indicates that the group averages are more spread out from the overall average.

2. Within-group variance (MSW):

It measures how much individual customer purchase values differ from their own age group's average.

A lower value indicates that customers within each age group have more similar purchase values.

Calculation steps:

First, we calculate the grand mean of all purchase values.

Then, we calculate the between-group sum of squares and divide it by its degrees of freedom.

Between-group variance = SSB / (Number of groups - 1)

Within-group variance = SSW / (Total observations - Number of groups)

In our dataset:
- Between-group variance = 65306.67
- Within-group variance = 690.00
The between-group variance is much higher than the within-group variance, indicating that the average purchase values differ considerably across the age groups.

Task 5: Perform One-Way ANOVA on Zomato customer segments

```python
import numpy as np
from scipy import stats

students = [1200, 1350, 1280, 1400, 1320]

working_professionals = [2200, 2400, 2300, 2500, 2350]

retirees = [1600, 1700, 1550, 1800, 1650]

print("Students Average:", np.mean(students))
print("Working Professionals Average:", np.mean(working_professionals))
print("Retirees Average:", np.mean(retirees))

f_statistic, p_value = stats.f_oneway(
    students,
    working_professionals,
    retirees
)
print("\\nANOVA Results")
print("F-statistic:", f_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("There is a significant difference between customer groups.")
else:
    print("There is no significant difference between customer groups.")
```
