```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Task 1: Calculate Percentiles of Zomato Delivery Times

```python
delivery_times = np.array([
    25, 30, 35, 28, 40,
    32, 45, 38, 29, 33,
    50, 42, 31, 36, 27,
    34, 39, 44, 30, 37
])

p25 = np.percentile(delivery_times, 25)
p50 = np.percentile(delivery_times, 50)
p75 = np.percentile(delivery_times, 75)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
```

Task 2: Calculate Quartiles of IPL Player Scores

```python
ipl_scores = np.array([
    12, 45, 67, 23, 89, 34,
    56, 78, 10, 95, 42, 61
])

q1 = np.quantile(ipl_scores, 0.25)
q2 = np.quantile(ipl_scores, 0.50)
q3 = np.quantile(ipl_scores, 0.75)

print("Q1 (First Quartile):", q1)
print("Q2 (Second Quartile):", q2)
print("Q3 (Third Quartile):", q3)
```

Task 3: Calculate IQR of Monthly Daily Steps

```python
steps = np.array([
    4500, 5200, 6100, 7000, 6500,
    8000, 7500, 6800, 7200, 9000,
    5500, 6200, 7100, 8300, 7600,
    6900, 6400, 7800, 8500, 5900,
    7300, 6700, 8100, 9200, 5600,
    6300, 7400, 7000, 6600, 8800
])

q1 = np.percentile(steps, 25)
q3 = np.percentile(steps, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("Interquartile Range:", iqr)
```

Explanation :

The IQR of daily steps is 1,425 steps. This means the middle 50% of daily step counts fall within a range of 1,425 steps. It indicates that the user's daily activity has relatively consistent variation during the month.

Remember:
- Q1 = 25th percentile
- Q2 = 50th percentile (Median)
- Q3 = 75th percentile
- IQR = Q3 − Q1

Task 4: Detect Outliers in Swiggy Order Amounts

```python
def detect_outliers(data):

    q1 = np.percentile(data, 25)
    q3 = np.percentile(data, 75)

    iqr = q3 - q1

    lower_limit = q1 - (1.5 * iqr)
    upper_limit = q3 + (1.5 * iqr)

    outliers = []

    for value in data:
        if value < lower_limit or value > upper_limit:
            outliers.append(value)

    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Outliers:", outliers)


# Swiggy order amounts
order_amounts = [
    250, 300, 350, 400, 450, 500,
    550, 600, 650, 700, 5000
]

detect_outliers(order_amounts)
```

Task 5: Create Boxplot of Weekly Screen Time

```python
screen_time = [4.5, 5.0, 5.2, 4.8, 6.0, 5.5, 4.9, 12.0]

plt.figure(figsize=(8, 5))

plt.boxplot(
    screen_time,
    flierprops=dict(marker='D', markersize=8)
)

plt.title("Weekly Screen Time Analysis")
plt.xlabel("Screen Time")
plt.ylabel("Hours")

plt.show()
```
