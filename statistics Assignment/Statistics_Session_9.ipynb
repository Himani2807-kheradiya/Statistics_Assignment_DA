```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
```

Task 1 – Generate a normal distribution and plot it

```python
np.random.seed(42)

data = np.random.normal(
    loc=500,
    scale=100,
    size=1000
)

print("First 10 Random Numbers:")
print(data[:10])

print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
```

```python
plt.figure(figsize=(10, 5))

plt.hist(data, bins=30, edgecolor='black')

plt.title("Normal Distribution of Sales Amounts")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")

plt.show()
```

Task 2 – Calculate Z-score for daily steps

```python
steps = np.array([6500, 7200, 8000, 8500, 9000,
                  9500, 10000, 7800, 8200, 8800])

x = 9500
mean = 8000
std_dev = 1200

z_score = (x - mean) / std_dev

print("Daily Steps:", x)
print("Mean Steps:", mean)
print("Standard Deviation:", std_dev)
print("Z-score:", z_score)
```

Task 3 – Standard normal curve with shaded area

```python
x = np.linspace(-4, 4, 1000)

y = norm.pdf(x, loc=0, scale=1)

plt.figure(figsize=(10, 5))

plt.plot(x, y, label="Standard Normal Curve")

x_fill = np.linspace(-1, 1, 500)
y_fill = norm.pdf(x_fill, loc=0, scale=1)

plt.fill_between(x_fill, y_fill, alpha=0.4)

plt.title("Standard Normal Distribution")
plt.xlabel("Z-score")
plt.ylabel("Probability Density")

plt.axvline(-1, linestyle='--')
plt.axvline(1, linestyle='--')

plt.legend()
plt.grid()
plt.show()
```

```python
probability = norm.cdf(1) - norm.cdf(-1)

print("Probability between Z = -1 and Z = 1:", probability)
print("Percentage:", probability * 100, "%")
```

Task 4 – Zomato delivery time probability

```python
mean = 30
std_dev = 5
delivery_time = 25

probability = norm.cdf(
    delivery_time,
    loc=mean,
    scale=std_dev
)

print("Mean Delivery Time:", mean)
print("Standard Deviation:", std_dev)
print("Delivery Time:", delivery_time)
print("Probability of delivery below 25 minutes:", probability)
print("Percentage:", probability * 100, "%")
```
