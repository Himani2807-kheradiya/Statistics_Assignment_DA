```python
import numpy as np
import pandas as pd
```

Task 1: Calculate Range of Daily Steps

```python
steps = np.array([6500, 7200, 8000, 6000, 9000, 7500, 8200])

maximum_steps = np.max(steps)
minimum_steps = np.min(steps)

step_range = maximum_steps - minimum_steps

print("Maximum Steps:", maximum_steps)
print("Minimum Steps:", minimum_steps)
print("Range of Steps:", step_range)
```

Explanation :

The range of daily steps is 3,000 steps. This means the difference between the user's most active day and least active day is 3,000 steps. It shows some variation in the user's daily activity.



Task 2: Calculate Variance and Standard Deviation of Zomato Orders

```python
order_amounts = np.array([250, 400, 350, 300, 500])

variance = np.var(order_amounts)
std_deviation = np.std(order_amounts)

print("Order Amounts:", order_amounts)
print("Mean:", np.mean(order_amounts))
print("Variance:", variance)
print("Standard Deviation:", std_deviation)
```

Explanation:

The variance is 8,500 and the standard deviation is approximately ₹92.20. This means the order amounts typically vary around the average order amount of ₹360.

Task 3: Calculate Coefficient of Variation (CV)

```python
data_usage = np.array([3.2, 2.8, 3.5, 3.0, 2.9, 3.1])

mean_usage = np.mean(data_usage)
std_usage = np.std(data_usage)

cv = (std_usage / mean_usage) * 100

print("Mean Usage:", mean_usage)
print("Standard Deviation:", std_usage)
print("Coefficient of Variation:", cv, "%")
```

Explanation :

The coefficient of variation is approximately 7.35%. This indicates that the user's monthly mobile data usage is quite consistent because the standard deviation is small compared to the mean.

Task 4: Instagram Likes – Standard Deviation vs Mean

```python
likes = np.array([120, 150, 130, 110, 170,
                  140, 135, 160, 125, 155])

mean_likes = np.mean(likes)
std_likes = np.std(likes)

cv_likes = (std_likes / mean_likes) * 100

print("Mean Likes:", mean_likes)
print("Standard Deviation:", std_likes)
print("Coefficient of Variation:", cv_likes, "%")
```

Explanation :

The average engagement is approximately 140 likes per post, with a standard deviation of 18.09 likes. The coefficient of variation is approximately 12.97%, which indicates relatively consistent engagement across the 10 Instagram posts.

Task 5: Generate UPI Transaction Dataset

```python
upi_transactions = np.array([
    120, 150, 100, 180,
    130, 160, 110, 140
])

upi_mean = np.mean(upi_transactions)
upi_variance = np.var(upi_transactions)
upi_std = np.std(upi_transactions)

print("UPI Transactions:", upi_transactions)
print("Mean:", upi_mean)
print("Variance:", upi_variance)
print("Standard Deviation:", upi_std)
```

Explanation :

The average UPI transaction amount is ₹136.25. The variance is 623.44 and the standard deviation is approximately ₹24.97. This indicates that transaction amounts have relatively low variation around the average, suggesting fairly stable spending in this example.
