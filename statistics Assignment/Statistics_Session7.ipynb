Task 1 – Probability of buying a phone given accessories

Given:
- P(Accessories) = 120 / 500
- P(Phone and Accessories) = 50 / 500
Formula:

P(\text{Phone}|\{Accessories}) =
\{P({Phone and Accessories})}
{P({Accessories})}
\]

```python
total_users = 500
accessories_users = 120
phone_users = 80
both_users = 50

prob_accessories = accessories_users / total_users
prob_both = both_users / total_users

prob_phone_given_accessories = prob_both / prob_accessories

print("Probability of Accessories:", prob_accessories)
print("Probability of Both:", prob_both)
print("P(Phone | Accessories):", prob_phone_given_accessories)
print("Percentage:", prob_phone_given_accessories * 100, "%")
```

Explanation:

Out of 120 users who bought phone accessories, 50 also bought a phone. Therefore, the probability of buying a phone given that the user already bought accessories is approximately 41.67%.

Task 3 – Bayes’ Theorem: Customer churn probability

```python
def bayes_churn_probability(prior_churn,
                            prob_email_given_churn,
                            prob_email):

    probability = (prior_churn * prob_email_given_churn) / prob_email

    return probability


result = bayes_churn_probability(
    prior_churn=0.15,
    prob_email_given_churn=0.6,
    prob_email=0.25
)

print("Probability of Churn:", result)
print("Percentage:", result * 100, "%")
```

EXPLANATION:
The probability that a user will churn, given that they received a "we miss you" email, is 36%. Bayes’ Theorem helps update the initial probability of churn using new information, such as receiving an email.

Task 4 – Bayes’ Theorem: Digital wallet fraud detection

```python
prob_fraud = 0.02
prob_alert_given_fraud = 0.90
prob_alert_given_normal = 0.05

prob_normal = 1 - prob_fraud

# Total probability of an alert
prob_alert = (
    prob_alert_given_fraud * prob_fraud
    + prob_alert_given_normal * prob_normal
)

# Probability of fraud given an alert
prob_fraud_given_alert = (
    prob_alert_given_fraud * prob_fraud
) / prob_alert

print("Probability of Alert:", prob_alert)
print("Probability of Fraud Given Alert:",
      prob_fraud_given_alert)
print("Percentage:",
      prob_fraud_given_alert * 100, "%")
```

EXPLANATION:The probability that a transaction is actually fraudulent when the alert system flags it is approximately 26.87%. Even though the system detects 90% of fraudulent transactions, many normal transactions can also be flagged. This is why the probability of actual fraud among flagged transactions is much lower than 90%.
