```python
import numpy as np
import pandas as pd
from statistics import mode
```

Task 1: Calculate Mean of Daily Steps

```python
steps = np.array([4500, 6000, 7500, 8000, 5500, 9000, 7000])

mean_steps = np.mean(steps)

print("Daily Steps:", steps)
print("Average Steps:", mean_steps)
```

Task 2: Calculate Median Delivery Time

```python
delivery_times = [32, 28, 29, 45, 30, 31, 60, 30, 29]

def find_median(data):
    return np.median(data)

median_time = find_median(delivery_times)

print("Delivery Times:", delivery_times)
print("Median Delivery Time:", median_time, "minutes")
```

Task 3: Find Mode of YouTube Video Genres

```python
genres = [
    'Music', 'Vlog', 'Music', 'Tech',
    'Music', 'Vlog', 'Tech', 'Music',
    'Comedy', 'Vlog', 'Music', 'Comedy'
]

most_watched = mode(genres)

print("Video Genres:", genres)
print("Most Watched Genre:", most_watched)
```

Task 4: Mean vs Median of UPI Transactions

```python
upi_amounts = np.array([100, 120, 105, 110, 5000, 115, 108])

mean_amount = np.mean(upi_amounts)
median_amount = np.median(upi_amounts)

print("UPI Transactions:", upi_amounts)
print("Mean Transaction Amount:", mean_amount)
print("Median Transaction Amount:", median_amount)
```

Explanation:

The median of ₹110 is a better measure of my typical UPI transaction because the ₹5,000 transaction is an unusually high value (outlier). It increases the mean to ₹808.29, making the average less representative of my normal spending. Therefore, the median gives a better idea of my usual transaction amount.
