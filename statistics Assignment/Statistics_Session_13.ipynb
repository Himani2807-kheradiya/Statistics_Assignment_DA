```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

Task 1 – One-sample t-test (Zomato delivery time)

```python
delivery_times = [28, 32, 27, 35, 29, 31, 30, 33, 26, 34]

print("Delivery Times:", delivery_times)
print("Sample Mean:", np.mean(delivery_times))
```

```python
t_statistic, p_value = stats.ttest_1samp(
    delivery_times,
    popmean=30
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
    print("Average delivery time is significantly different from 30 minutes.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("Average delivery time is not significantly different from 30 minutes.")
```

Task 2 – Independent (Two-sample) t-test

```python
before_steps = [4500, 5000, 5200, 4800, 5500,
                4700, 5100, 4900, 5300, 4600]

after_steps = [6000, 6500, 6200, 6100, 7000,
               5900, 6800, 6300, 6600, 6100]

print("Average Steps Before:", np.mean(before_steps))
print("Average Steps After:", np.mean(after_steps))
```

```python
t_statistic, p_value = stats.ttest_ind(
    before_steps,
    after_steps
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
    print("Daily steps increased significantly after using the app.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("No significant increase in daily steps.")
```

Task 3 – Paired t-test (Focus Mode)

```python
before_screen = [6.5, 7, 5.5, 6, 8,
                 7.5, 6.5, 7, 5, 6]

after_screen = [5.5, 6, 4.5, 5, 6.5,
                6, 5, 5.5, 4, 5]

print("Average Screen Time Before:", np.mean(before_screen))
print("Average Screen Time After:", np.mean(after_screen))
```

```python
t_statistic, p_value = stats.ttest_rel(
    before_screen,
    after_screen
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
    print("Focus Mode significantly reduced screen time.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("No significant reduction in screen time.")
```

Task 4 – Compare Zomato and Swiggy ratings

```python
zomato_ratings = [
    4.5, 4.2, 4.8, 4.0, 4.6,
    4.3, 4.7, 4.1, 4.4, 4.9,
    4.2, 4.5, 4.6, 4.3, 4.7
]

swiggy_ratings = [
    4.2, 4.0, 4.5, 4.1, 4.3,
    4.4, 4.6, 3.9, 4.2, 4.1,
    4.5, 4.0, 4.3, 4.2, 4.4
]

print("Average Zomato Rating:", np.mean(zomato_ratings))
print("Average Swiggy Rating:", np.mean(swiggy_ratings))
```

```python
t_statistic, p_value = stats.ttest_ind(
    zomato_ratings,
    swiggy_ratings,
    equal_var=False
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
    print("There is a significant difference in app ratings.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no significant difference in app ratings.")
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    [zomato_ratings, swiggy_ratings],
    tick_labels=['Zomato', 'Swiggy']
)

plt.title("Zomato vs Swiggy Ratings")
plt.ylabel("User Ratings")
plt.show()
```

An independent two-sample t-test is appropriate because Zomato and Swiggy ratings are collected from two different groups of users. We use Welch's t-test because it does not assume equal population variances. The p-value is approximately 0.0253, which is below 0.05. Therefore, we reject the null hypothesis and conclude that there is a statistically significant difference between the average ratings in this simulated dataset. Zomato has the higher sample average rating.

Task 5 – One-sample t-test using UPI transactions

```python
# Daily UPI transactions per user
upi_transactions = [
    18, 22, 20, 25, 19, 24,
    21, 23, 17, 26, 22, 20
]

print("UPI Transactions:", upi_transactions)
print("Average Transactions:", np.mean(upi_transactions))
```

```python
# Assumed population average
expected_average = 20

t_statistic, p_value = stats.ttest_1samp(
    upi_transactions,
    popmean=expected_average
)

print("Expected Average:", expected_average)
print("Sample Average:", np.mean(upi_transactions))
print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
    print("Average UPI transactions are significantly different.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("Average UPI transactions are not significantly different.")
```

Explanation :

A one-sample t-test compares the average of one sample with an assumed population average. Here, we compare the average daily UPI transactions of 12 users with the assumed average of 20 transactions. The p-value is approximately 0.105, which is greater than 0.05. Therefore, we fail to reject the null hypothesis. There is not enough statistical evidence to conclude that the average daily UPI transactions are different from 20.
