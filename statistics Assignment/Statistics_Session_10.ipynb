Task 2 – Stratified sampling: Flipkart orders

```python
# Create 1000 Flipkart orders
cities = (
    ['Ahmedabad'] * 500 +
    ['Surat'] * 300 +
    ['Vadodara'] * 200
)

orders = pd.DataFrame({
    'Order_ID': range(1, 1001),
    'City': cities
})

print(orders.head())
print("\nCity-wise order count:")
print(orders['City'].value_counts())
```

```python
np.random.seed(42)

sample_size = 60
stratified_sample = []

for city in orders['City'].unique():

    city_orders = orders[orders['City'] == city]

    city_sample_size = round(
        len(city_orders) / len(orders) * sample_size
    )

    selected_orders = city_orders.sample(
        n=city_sample_size,
        replace=False,
        random_state=42
    )

    stratified_sample.append(selected_orders)

final_sample = pd.concat(stratified_sample)

print("Stratified Sample:")
print(final_sample)

print("\nCity-wise sample count:")
print(final_sample['City'].value_counts())

print("\nTotal Sample Size:", len(final_sample))
```

Task 4 – Importance of sample size: Swiggy delivery times

```python
population_size = 10000
population_std = 10

sample_sizes = [50, 200, 500]

results = []

for n in sample_sizes:

    standard_error = (
        population_std / np.sqrt(n)
    ) * np.sqrt(
        (population_size - n) / (population_size - 1)
    )

    margin_of_error = 1.96 * standard_error

    results.append({
        'Sample Size': n,
        'Standard Error': standard_error,
        'Margin of Error': margin_of_error
    })

result_df = pd.DataFrame(results)

print(result_df.round(3))
```

- 50 records: The estimate has more uncertainty and a larger margin of error.
- 200 records: The estimate becomes more reliable because the sample size increases.
- 500 records: The estimate has the smallest margin of error among the three options.
Therefore, a sample of 500 records is expected to give a more accurate estimate than samples of 50 or 200 records, assuming the records are selected randomly and representatively.

Task 5 – Stratified sampling: IPL cricket fans

```python
# Create a dataset of IPL fans
teams = (
    ['CSK'] * 60 +
    ['MI'] * 50 +
    ['RCB'] * 40 +
    ['KKR'] * 30 +
    ['GT'] * 20
)

fans = pd.DataFrame({
    'Fan_ID': range(1, 201),
    'Favorite_Team': teams
})

print("IPL Fan Dataset:")
print(fans.head())

print("\nTeam-wise Fan Count:")
print(fans['Favorite_Team'].value_counts())
```

```python
np.random.seed(42)

target_sample_size = 40
selected_fans = []

team_groups = fans.groupby('Favorite_Team')

for team, group in team_groups:

    proportion = len(group) / len(fans)

    team_sample_size = round(
        proportion * target_sample_size
    )

    chosen = group.sample(
        n=team_sample_size,
        replace=False,
        random_state=42
    )

    selected_fans.append(chosen)

ipl_sample = pd.concat(selected_fans)

print("Selected IPL Fans:")
print(ipl_sample)

print("\nStratified Sample by Team:")
print(ipl_sample['Favorite_Team'].value_counts())

print("\nTotal Sample Size:", len(ipl_sample))
```
