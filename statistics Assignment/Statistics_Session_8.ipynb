```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson
```

Task 1 – Bernoulli trial: Zomato login

```python
p = 0.85


results = np.random.binomial(1, p, 20)

print("Login Results:")
print(results)

print("Total Successful Logins:", np.sum(results))
print("Total Failed Logins:", 20 - np.sum(results))
```

Task 2 – Binomial distribution: Paytm UPI payments

```python
def payment_probability(n, k, p):
    probability = binom.pmf(k, n, p)
    return probability


result = payment_probability(5, 3, 0.9)

print("Probability of Exactly 3 Successful Payments:", result)
print("Percentage:", result * 100, "%")
```

Task 3 – Poisson distribution: Spotify playlist creation

```python
average_rate = 4
number_of_playlists = 6

probability = poisson.pmf(number_of_playlists, average_rate)

print("Average Playlist Creations:", average_rate)
print("Number of Playlists:", number_of_playlists)
print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

Task 4 – Poisson simulation: Flipkart product reviews

```python
average_reviews = 10

np.random.seed(42)

reviews = np.random.poisson(
    lam=average_reviews,
    size=12
)

print("Reviews received in 12 hours:")
print(reviews)

print("Total Reviews:", np.sum(reviews))
print("Average Reviews:", np.mean(reviews))
```

```python
hours = range(1, 13)

plt.figure(figsize=(10, 5))

plt.bar(hours, reviews)

plt.title("Flipkart Product Reviews Per Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Reviews")

plt.xticks(hours)
plt.grid(axis='y')

plt.show()
```

Task 5 – Real-world analytics use-case: Swiggy

Selected distribution: Poisson Distribution

Scenario: Swiggy – Predicting incoming food orders

Swiggy can use the Poisson distribution to estimate the number of food orders a restaurant may receive within a fixed time period, such as 10 minutes.

For example, suppose a restaurant receives an average of 8 orders every 10 minutes. Swiggy can use the Poisson distribution to estimate the probability of receiving 5, 8, or 12 orders during the next 10 minutes.

This information can help restaurants prepare ingredients, manage kitchen staff, estimate delivery requirements, and reduce customer waiting time.

Why this distribution fits (my own explanation):

The Poisson distribution is suitable because it models the number of events occurring in a fixed time interval when the average event rate is known. Food orders arrive over time, so it can help Swiggy and restaurants plan their resources more efficiently.
