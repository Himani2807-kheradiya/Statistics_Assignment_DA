Task 4: Automatic Measurement Scale Detection

```python
def detect_scale(column_name, values):

    name = column_name.lower()

    nominal_words = ['name', 'city', 'status', 'category',
                     'payment', 'method', 'id', 'type']

    if any(word in name for word in nominal_words):
        return "Nominal"

    ordinal_words = ['rating', 'rank', 'level',
                     'priority', 'grade', 'review']

    if any(word in name for word in ordinal_words):
        return "Ordinal"

    interval_words = ['temperature', 'date', 'time',
                      'year', 'celsius']

    if any(word in name for word in interval_words):
        return "Interval"

    ratio_words = ['price', 'amount', 'cost', 'count',
                   'quantity', 'likes', 'followers',
                   'duration', 'weight', 'age']

    if any(word in name for word in ratio_words):
        return "Ratio"

    
    if all(isinstance(x, (int, float)) for x in values):
        return "Likely Ratio - Needs Review"

    return "Nominal"
```

```python
print(detect_scale('Payment Method', ['UPI', 'Cash', 'Card']))

print(detect_scale('Product Rating', [3, 4, 5, 2]))

print(detect_scale('Product Price', [500, 700, 900]))

print(detect_scale('Temperature', [25, 30, 35]))

print(detect_scale('Number of Likes', [100, 200, 300]))
```

Task 4 – Explanation
The function uses simple rules based on column names and value types.
1. Nominal: Identifies names, cities, payment methods and statuses.
2. Ordinal: Identifies ratings, rankings, grades and priority levels.
3. Interval: Identifies temperature, dates and time-related columns.
4. Ratio: Identifies prices, quantities, counts and amounts.
