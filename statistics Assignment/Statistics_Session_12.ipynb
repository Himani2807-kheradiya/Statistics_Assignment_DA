```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

Task 1 – State hypotheses using Python

```python
print("Null Hypothesis (H0): Cashback offer does not increase average daily orders.")
print("Alternative Hypothesis (H1): Cashback offer increases average daily orders.")
```

Task 2 – Identify Type I and Type II errors

```python
print("Type I Error: Concluding that the algorithm works when it actually does not.")
print("Type II Error: Concluding that the algorithm does not work when it actually works.")
```

Task 3 – Test p-value (Flipkart)

```python
p_value = 0.03
confidence_level = 0.95

alpha = 1 - confidence_level

print("P-value:", p_value)
print("Significance Level:", alpha)

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("The new discount banner significantly increases purchases.")
else:
    print("Fail to Reject the Null Hypothesis")
```

Task 4 – 99% Confidence Level

```python
confidence_level = 0.99

alpha = 1 - confidence_level

p_value = 0.008

print("Confidence Level:", confidence_level * 100, "%")
print("Significance Level:", alpha)
print("P-value:", p_value)

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is significant evidence of increased user engagement.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is not enough evidence of increased user engagement.")
```

Bonus – Visualize p-value decisions

```python
import matplotlib.pyplot as plt

p_values = [0.03, 0.008, 0.07]
labels = ['Flipkart (95%)', 'Cricket App (99%)', 'Example (95%)']
alpha_values = [0.05, 0.01, 0.05]

for i in range(len(p_values)):
    if p_values[i] < alpha_values[i]:
        print(labels[i], ": Reject H0")
    else:
        print(labels[i], ": Fail to Reject H0")

plt.figure(figsize=(8, 5))
plt.bar(labels, p_values)
plt.axhline(y=0.05, linestyle='--', label='Alpha = 0.05')
plt.title("P-value Comparison")
plt.ylabel("P-value")
plt.legend()
plt.show()
```
