```python
import pandas as pd
import numpy as np
from scipy.stats import spearmanr
import matplotlib.pyplot as plt
import seaborn as sns
```

Task 1: Calculate Pearson correlation between danceability and popularity

```python
data = {
    "song": [
        "Song A", "Song B", "Song C", "Song D",
        "Song E", "Song F", "Song G", "Song H",
        "Song I", "Song J", "Song K", "Song L"
    ],
    "danceability": [
        0.82, 0.75, 0.68, 0.90, 0.55, 0.72,
        0.60, 0.88, 0.47, 0.79, 0.64, 0.95
    ],
    "energy": [
        0.70, 0.85, 0.60, 0.92, 0.45, 0.78,
        0.52, 0.89, 0.40, 0.81, 0.66, 0.95
    ],
    "popularity": [
        85, 78, 65, 92, 50, 74,
        55, 88, 42, 80, 62, 95
    ]
}

songs = pd.DataFrame(data)

songs.to_csv("song_data.csv", index=False)

print(songs)
```

```python
songs = pd.read_csv("song_data.csv")

pearson_corr = songs["danceability"].corr(
    songs["popularity"],
    method="pearson"
)

print("Pearson Correlation:", pearson_corr)
```

Task 2: Calculate Spearman correlation between IPL players' runs and Instagram followers

```python
players = pd.DataFrame({
    "player": [
        "Player A", "Player B", "Player C", "Player D", "Player E",
        "Player F", "Player G", "Player H", "Player I", "Player J"
    ],
    "total_runs": [
        850, 720, 650, 580, 510,
        430, 350, 280, 190, 120
    ],
    "instagram_followers": [
        25, 18, 22, 12, 15,
        9, 11, 5, 7, 3
    ]
})

print(players)

correlation, p_value = spearmanr(
    players["total_runs"],
    players["instagram_followers"]
)

print("\nSpearman Correlation:", correlation)
print("P-value:", p_value)
```

Task 3: Create a correlation matrix for Flipkart product features

```python
products = pd.DataFrame({
    "price": [
        499, 799, 299, 1299, 999,
        399, 1599, 699, 899, 199
    ],
    "user_rating": [
        4.2, 4.5, 3.9, 4.7, 4.4,
        4.0, 4.8, 4.3, 4.6, 3.8
    ],
    "number_of_reviews": [
        120, 350, 80, 900, 500,
        95, 1200, 260, 650, 45
    ]
})

print(products)

corr_matrix = products.corr()

print("\nCorrelation Matrix:")
print(corr_matrix)
```

Task 4: Visualize the correlation matrix using a heatmap

```python
corr_matrix = products.corr()

plt.figure(figsize=(8, 5))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    cbar=True,
    linewidths=0.5,
    fmt=".2f"
)

plt.title("Flipkart Product Features - Correlation Heatmap")



plt.show()
```

Task 5: Two potential pitfalls when interpreting correlation

1. Correlation does not mean causation

Correlation between two variables does not prove that one variable causes the other.

Example: On Zomato, we may find that restaurants with more customer reviews also have more orders. However, more reviews do not necessarily cause more orders. Restaurant popularity, location, and food quality may influence both.

2. Influence of a third variable (confounding variable)

Sometimes, two variables appear related because another variable influences both of them.

Example: On YouTube, we may find that videos with more views also have more likes. However, video promotion or a large subscriber base may increase both views and likes. Therefore, views alone may not explain the increase in likes.
