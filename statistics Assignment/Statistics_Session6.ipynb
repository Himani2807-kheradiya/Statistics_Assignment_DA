Task 1 – Sample space: Die and coin

Question: List all possible outcomes when rolling a six-sided die and flipping a coin at the same time.

A standard die has 6 possible outcomes: 1, 2, 3, 4, 5, 6.

A coin has 2 possible outcomes: Heads (H) and Tails (T).

```python
die = [1, 2, 3, 4, 5, 6]
coin = ['Heads', 'Tails']

sample_space = []

for number in die:
    for side in coin:
        sample_space.append((number, side))

print("Sample Space:")
print(sample_space)

print("Total Outcomes:", len(sample_space))
```

Task 2 – Classical probability of an even number

Question: A deck has 10 cards numbered from 1 to 10. Find the probability of drawing an even number.

Even numbers are: 2, 4, 6, 8, 10.

Formula:
\[
P(A)={Favourable Outcomes}/{Total Outcomes}
\]

```python
cards = list(range(1, 11))

even_cards = []

for card in cards:
    if card % 2 == 0:
        even_cards.append(card)

probability = len(even_cards) / len(cards)

print("All Cards:", cards)
print("Even Cards:", even_cards)
print("Favourable Outcomes:", len(even_cards))
print("Total Outcomes:", len(cards))
print("Probability:", probability)
```

Task 3 – Probability of NOT selecting an Arijit Singh song

```python
total_songs = 8
arijit_songs = 3

other_songs = total_songs - arijit_songs

probability = other_songs / total_songs

print("Total Songs:", total_songs)
print("Arijit Singh Songs:", arijit_songs)
print("Songs by Other Artists:", other_songs)
print("Probability of NOT selecting an Arijit Singh song:", probability)
```

Explanation :
Out of 8 songs, 5 songs are by other artists. Therefore, the probability of randomly selecting a song that is not by Arijit Singh is 0.625 or 62.5%.

Task 4 – Addition rule: Pizza or Burger

```python
p_pizza = 0.3
p_burger = 0.4
p_both = 0.1

p_pizza_or_burger = p_pizza + p_burger - p_both

print("Probability of Pizza:", p_pizza)
print("Probability of Burger:", p_burger)
print("Probability of Both:", p_both)
print("Probability of Pizza or Burger:", p_pizza_or_burger)
```

Explanation:

The probability of ordering Pizza or Burger is 0.6 or 60%. We subtract the probability of ordering both because those customers are counted twice when adding the two individual probabilities.

Task 5 – Conditional probability: Flipkart

```python
p_phone = 0.2
p_phone_given_mobile_interest = 0.5

print("Probability of adding a phone to cart:", p_phone)
print("Probability given mobile interest:", p_phone_given_mobile_interest)
```

Explanation:

Conditional probability means the probability of an event happening when we already know that another event has happened or a particular condition is true.

In this example, the probability of a random Flipkart user adding a phone to their cart is 0.2 (20%). However, if we know that the user is interested in mobiles, the probability increases to 0.5 (50%). This means mobile-interest users are more likely to add a phone to their cart.

Note: The 0.5 is the probability given mobile interest, not the probability of being interested in mobiles.
