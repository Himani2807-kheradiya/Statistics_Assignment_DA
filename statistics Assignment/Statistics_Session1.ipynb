TASK 1:random numbers between 100 and 1000 to simulate Zomato order values for a day.

```python
import numpy as np
import pandas as pd
from statistics import mode
```

```python
order_amount = [250, 450, 300, 500, 250,
                700, 350, 450, 600, 250,
                800, 400, 550, 300, 450,
                900, 350, 500, 250, 650]

df = pd.DataFrame(order_amount, columns=['Order Amount'])

df
```

TASK 2: Calculate Mean, Median and Mode


```python
mean_value = df['Order Amount'].mean()
median_value = df['Order Amount'].median()
mode_value = df['Order Amount'].mode()[0]

print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

TASK 3: Calculate Sample Mean 

```python
sample = df.head(5)

print("Sample Data:")
print(sample)

sample_mean = sample['Order Amount'].mean()
population_mean = df['Order Amount'].mean()

print("Sample Mean:", sample_mean)
print("Population Mean:", population_mean)
print("Difference:", population_mean - sample_mean)
```

TASK 4: Population vs Sample Explanation 

Population: Population refers to the complete set of data we want to study. In this example, all 20 Zomato orders received in one day represent the population.

Sample: A sample is a smaller part selected from the population. In this example, the first 5 orders out of 20 represent the sample.

Which is more practical?

For a food delivery company like Zomato, using a sample is often more practical because the company receives thousands of orders every day. Analyzing a representative sample saves time and resources while helping the company understand customer preferences and ordering patterns.
