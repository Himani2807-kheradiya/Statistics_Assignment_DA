```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2_contingency, chisquare
```

Task 1 – Create a contingency table

```python

data = {
    'App': [
        'Zomato', 'Zomato', 'Zomato', 'Zomato', 'Zomato',
        'Zomato', 'Zomato', 'Zomato', 'Zomato', 'Zomato',
        'Swiggy', 'Swiggy', 'Swiggy', 'Swiggy', 'Swiggy',
        'Swiggy', 'Swiggy', 'Swiggy', 'Swiggy', 'Swiggy',
        'Uber Eats', 'Uber Eats', 'Uber Eats', 'Uber Eats', 'Uber Eats',
        'Uber Eats', 'Uber Eats', 'Uber Eats', 'Uber Eats', 'Uber Eats'
    ],

    'Payment': [
        'Paytm', 'Paytm', 'Paytm', 'PhonePe', 'PhonePe',
        'PhonePe', 'PhonePe', 'Credit Card', 'Credit Card', 'Paytm',
        'Paytm', 'Paytm', 'PhonePe', 'PhonePe', 'PhonePe',
        'Credit Card', 'Credit Card', 'Credit Card', 'Paytm', 'PhonePe',
        'Paytm', 'PhonePe', 'PhonePe', 'Credit Card', 'Credit Card',
        'Credit Card', 'Paytm', 'PhonePe', 'Credit Card', 'Paytm'
    ]
}

df = pd.DataFrame(data)

print(df.head(10))
```

```python
contingency_table = pd.crosstab(
    df['App'],
    df['Payment']
)

print("Contingency Table:")
print(contingency_table)
```

```python
print("Row Totals:")
print(contingency_table.sum(axis=1))

print("\nColumn Totals:")
print(contingency_table.sum(axis=0))

print("\nGrand Total:")
print(contingency_table.values.sum())
```

Task 2 – Calculate expected counts

```python
observed = np.array([
    [12, 18],
    [8, 7],
    [10, 15]
])

music_table = pd.DataFrame(
    observed,
    index=['Spotify', 'Gaana', 'YouTube Music'],
    columns=['Males', 'Females']
)

print(music_table)
```

```python
row_totals = observed.sum(axis=1)
column_totals = observed.sum(axis=0)
grand_total = observed.sum()

print("Row Totals:", row_totals)
print("Column Totals:", column_totals)
print("Grand Total:", grand_total)
```

```python
expected = np.outer(
    row_totals,
    column_totals
) / grand_total

expected_table = pd.DataFrame(
    expected,
    index=['Spotify', 'Gaana', 'YouTube Music'],
    columns=['Males', 'Females']
)

print("Expected Counts:")
print(expected_table.round(2))
```

Task 3 – Chi-square test of independence

```python
print(contingency_table)
```

```python
chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("Chi-square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)

print("\nExpected Frequencies:")
print(pd.DataFrame(
    expected,
    index=contingency_table.index,
    columns=contingency_table.columns
).round(2))
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("There is a significant association between app choice and payment method.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("There is no significant association between app choice and payment method.")
```

Step 4: Show calculation steps

The chi-square statistic is calculated by comparing observed and expected frequencies.
The Chi-Square statistic is calculated using:

$$
\chi^2 = \sum \frac{(O-E)^2}{E}
$$

Where:
- O = Observed frequency
- E = Expected frequency
Degrees of freedom:
$$
df=(r-1)(c-1)
$$

Here:
$$
df=(3-1)(3-1)=4
$$

Explanation :

The chi-square test of independence checks whether two categorical variables are related. Here, we test whether favorite food delivery app and preferred payment method are associated. The p-value is approximately 0.8781, which is greater than 0.05. Therefore, we fail to reject the null hypothesis. There is not enough statistical evidence to conclude that app choice and payment method are associated in this sample.

Task 4 – Chi-square goodness of fit test

```python
observed = np.array([18, 12])
claimed_proportions = np.array([0.70, 0.30])

total_users = 30

expected = total_users * claimed_proportions

print("Observed Counts:", observed)
print("Expected Counts:", expected)
```

```python
chi2, p_value = chisquare(
    f_obs=observed,
    f_exp=expected
)

print("Chi-square Statistic:", chi2)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
    print("The observed preference does not fit the company's claim.")
else:
    print("Fail to Reject the Null Hypothesis")
    print("The observed preference is consistent with the company's claim.")
```

Calculation steps:
Category	Observed	Expected
Prefer product	18	21
Do not prefer	12	9

$$
\chi^2 = \frac{(18-21)^2}{21}+\frac{(12-9)^2}{9}
$$

$$
\chi^2 = 0.4286+1=1.4286
$$

Degrees of freedom:
$$
df=2-1=1
$$

Explanation :

The chi-square goodness of fit test checks whether observed data matches an expected distribution. Here, Flipkart claims that 70% of users will prefer its new product. For 30 users, the expected counts are 21 who prefer it and 9 who do not. The p-value is approximately 0.232, which is greater than 0.05. Therefore, we fail to reject the null hypothesis. There is not enough evidence to say that the observed results differ from the company's 70% claim.
