```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Task 1 – Simulate the Central Limit Theorem

```python
np.random.seed(42)

likes = np.random.lognormal(
    mean=3,
    sigma=1,
    size=10000
)

print(likes[:10])
```

```python
sample_means = []

for i in range(1000):
    sample = np.random.choice(likes, size=30)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)

print("Number of sample means:", len(sample_means))
print("Average of sample means:", np.mean(sample_means))
```

```python
plt.figure(figsize=(9, 5))

plt.hist(sample_means, bins=30, edgecolor='black')

plt.title("Central Limit Theorem - Sample Means (n=30)")
plt.xlabel("Sample Means of Instagram Likes")
plt.ylabel("Frequency")

plt.show()
```

Task 2 – Explain why the sampling distribution becomes normal

The Central Limit Theorem explains that when we repeatedly take random samples from a population and calculate their averages, the distribution of those averages becomes approximately normal, even if the original data is not normally distributed.

For example, consider Zomato order amounts. Some customers may place small orders, while others may place very expensive orders. The original order amounts may be skewed. However, if we repeatedly select random groups of 30 orders and calculate their average amounts, the distribution of those averages will become approximately bell-shaped.

How CLT works

                                        Original population
                                        Skewed order amounts
                                                |
        Sample 1                            Sample 2                         Sample 3
        Mean ₹250                           Mean ₹280                        Mean ₹265
                                                |
                                            Sampling distribution
                                            Approximately bell-shaped

Task 4 – Compare sample size n=5 and n=30

```python
means_5 = []

for i in range(1000):
    sample = np.random.choice(likes, size=5)
    means_5.append(np.mean(sample))


means_30 = []

for i in range(1000):
    sample = np.random.choice(likes, size=30)
    means_30.append(np.mean(sample))


plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(means_5, bins=30, edgecolor='black')
plt.title("Sample Means (n=5)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(means_30, bins=30, edgecolor='black')
plt.title("Sample Means (n=30)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

When the sample size is 5, the histogram of sample means may still look skewed and spread out. When the sample size increases to 30, the histogram becomes more bell-shaped and narrower.

This happens because larger samples produce more stable averages. Therefore, increasing the sample size improves the normal approximation and reduces the spread of the sampling distribution.

Task 5 – CLT using an Exponential Distribution

```python
np.random.seed(42)

data = np.random.exponential(
    scale=2,
    size=10000
)

plt.figure(figsize=(8, 5))

plt.hist(data, bins=50, edgecolor='black')

plt.title("Original Exponential Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
sample_means_exp = []

for i in range(1000):
    sample = np.random.choice(data, size=30)
    sample_means_exp.append(np.mean(sample))

print("Number of sample means:", len(sample_means_exp))
print("Mean of sample means:", np.mean(sample_means_exp))
```

```python
plt.figure(figsize=(9, 5))

plt.hist(
    sample_means_exp,
    bins=30,
    edgecolor='black'
)

plt.title("CLT Using Exponential Distribution")
plt.xlabel("Sample Means")
plt.ylabel("Frequency")

plt.show()
```

The original exponential distribution is right-skewed and does not have a normal shape. After taking 1,000 random samples of size 30 and calculating their means, the histogram of sample means becomes approximately bell-shaped. This demonstrates the Central Limit Theorem. It shows that the sampling distribution of the mean can become approximately normal even when the original population is not normally distributed.
